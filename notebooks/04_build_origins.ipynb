```python
# Bibliotecas
import geopandas as gpd
```

```python
# Grade do IBGE
grafo = gpd.read_file("../data/grade_estatistica-2022/grade_id14.shp")
print(grafo.columns)
grafo.info()
```

```python
# Grade do IBGE
grafo = gpd.read_file("../data/grade_estatistica_5k-2022/BR5KM.shp")
print(grafo.columns)
# print(grafo.columns.tolist())
grafo.info()
```

```python
# Apenas células habitadas
origins = grafo[grafo["TOTAL"] > 0].copy()

# Centróide da célula
origins["geometry"] = origins.geometry.centroid

print(origins["geometry"].head())

```

```python
print(f"Número de células habitadas: {len(origins)}")
```

```python
# Verifica o sistema de coordenadas
print(origins.crs)
```

```python
# Converte para o sistema de coordenadas desejado
print(origins["geometry"].head().to_crs(epsg=31982))  
```
