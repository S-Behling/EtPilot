```python
# ============================================================
# 1. DIRETÓRIOS DO PROJETO
# ============================================================

# Identifica a raiz do projeto

CURRENT_DIR = Path.cwd()

if CURRENT_DIR.name == "notebooks":
    PROJECT_ROOT = CURRENT_DIR.parent
else:
    PROJECT_ROOT = CURRENT_DIR


# Cria uma pasta específica para os dados do Censo 2022.

CENSO_DIR = (
    PROJECT_ROOT
    / "data"
    / "censo_2022"
)


MICRODADOS_DIR = (
    CENSO_DIR
    / "microdados_publicos"
)


APOND_DIR = (
    CENSO_DIR
    / "areas_ponderacao"
)


DOCUMENTACAO_DIR = (
    CENSO_DIR
    / "dicionarios"
)


# Cria as pastas caso ainda não existam

for pasta in [
    CENSO_DIR,
    MICRODADOS_DIR,
    APOND_DIR,
    DOCUMENTACAO_DIR
]:
    pasta.mkdir(
        parents=True,
        exist_ok=True
    )


print(CENSO_DIR)
```

```python
# ============================================================
# 3. MICRODADOS PÚBLICOS - RIO GRANDE DO SUL
# ============================================================

# Somente os dados do Censo 2022

URL_MICRODADOS_RS = (
    "https://ftp.ibge.gov.br/"
    "Censos/Censo_Demografico_2022/"
    "Microdados_e_Areas_de_Ponderacao/"
    "Microdados_de_acesso_Publico/"
    "csv/"
    "43_RS.zip"
)


MICRODADOS_ZIP = (
    MICRODADOS_DIR
    / "43_RS.zip"
)


dwl.baixar_arquivo(URL_MICRODADOS_RS, MICRODADOS_ZIP)
```

```python
# ============================================================
# 4. EXTRAÇÃO DOS MICRODADOS
# ============================================================

MICRODADOS_EXTRAIDOS = (
    MICRODADOS_DIR
    / "43_RS"
)


MICRODADOS_EXTRAIDOS.mkdir(
    parents=True,
    exist_ok=True
)


with zipfile.ZipFile(
    MICRODADOS_ZIP,
    "r"
) as zip_ref:

    zip_ref.extractall(
        MICRODADOS_EXTRAIDOS
    )


print(
    "Microdados extraídos em:"
)

print(
    MICRODADOS_EXTRAIDOS
)
```

```python
for arquivo in MICRODADOS_EXTRAIDOS.rglob("*"):

    if arquivo.is_file():

        print(
            arquivo.name
        )
```
